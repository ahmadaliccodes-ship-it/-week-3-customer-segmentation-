```python
from pathlib import Path
import os
PROJECT_ROOT = Path.cwd() if (Path.cwd() / "data").exists() else Path.cwd().parent
os.chdir(PROJECT_ROOT)
print(f"Project root: {PROJECT_ROOT}")
```

```python
import pandas as pd

# ---------------------------------------------------------------------------
# Step 1: Project Objective
# ---------------------------------------------------------------------------
PROJECT_OBJECTIVE = (
    "Segment wholesale distribution customers into meaningful groups based on "
    "their annual spending across six product categories (Fresh, Milk, Grocery, "
    "Frozen, Detergents_Paper, Delicassen), using unsupervised clustering "
    "(K-Means, cross-checked with Hierarchical Clustering), in order to reveal "
    "actionable customer archetypes for targeted business decisions "
    "(e.g. promotions, delivery/route planning, inventory allocation)."
)

DATA_PATH = "data/raw/wholesale_customers.csv"

# ---------------------------------------------------------------------------
# Step 2: Load the dataset
# ---------------------------------------------------------------------------
df_raw = pd.read_csv(DATA_PATH)

# ---------------------------------------------------------------------------
# Step 3: Preserve the raw data
# df_raw is never modified beyond this point. All downstream phases must
# create a copy (e.g. df = df_raw.copy()) before any transformation.
# A frozen CSV snapshot is also written to disk for auditability.
# ---------------------------------------------------------------------------
df_raw.to_csv("df_raw_snapshot.csv", index=False)

# ---------------------------------------------------------------------------
# Step 4: Understand the dataset
# ---------------------------------------------------------------------------
def profile_dataset(df: pd.DataFrame) -> dict:
    """Return a structured profile of the dataset for the report's
    'Introduction & Dataset Overview' section."""
    profile = {
        "shape": df.shape,
        "columns": list(df.columns),
        "dtypes": df.dtypes.astype(str).to_dict(),
        "missing_values": df.isnull().sum().to_dict(),
        "duplicate_rows": int(df.duplicated().sum()),
        "n_unique_channel": df["Channel"].nunique(),
        "n_unique_region": df["Region"].nunique(),
        "channel_counts": df["Channel"].value_counts().to_dict(),
        "region_counts": df["Region"].value_counts().to_dict(),
    }
    return profile

DATA_DICTIONARY = {
    "Channel": "Distribution channel: 1 = Horeca (Hotel/Restaurant/Cafe), 2 = Retail.",
    "Region": "Geographic region code: 1 = Lisbon, 2 = Oporto, 3 = Other region.",
    "Fresh": "Annual spending in monetary units on fresh products.",
    "Milk": "Annual spending in monetary units on milk products.",
    "Grocery": "Annual spending in monetary units on grocery products.",
    "Frozen": "Annual spending in monetary units on frozen products.",
    "Detergents_Paper": "Annual spending in monetary units on detergents and paper products.",
    "Delicassen": "Annual spending in monetary units on delicatessen products.",
}


if __name__ == "__main__":
    print("=" * 70)
    print("PROJECT OBJECTIVE")
    print("=" * 70)
    print(PROJECT_OBJECTIVE)

    print("\n" + "=" * 70)
    print("DATASET PROFILE")
    print("=" * 70)

    profile = profile_dataset(df_raw)
    print(f"Shape (rows, columns): {profile['shape']}")
    print(f"Columns: {profile['columns']}")
    print(f"\nDtypes:\n{pd.Series(profile['dtypes'])}")
    print("\nData dictionary:")
    for col in df_raw.columns:
        print(f"- {col}: {DATA_DICTIONARY.get(col, 'No description available')}")
    print(f"\nMissing values per column:\n{pd.Series(profile['missing_values'])}")
    print(f"\nDuplicate rows: {profile['duplicate_rows']}")
    print(f"\nChannel value counts: {profile['channel_counts']}")
    print(f"Region value counts: {profile['region_counts']}")

    print("\nFirst 5 rows (df_raw.head()):")
    print(df_raw.head().to_string())

    print("\nSummary statistics (df_raw.describe()):")
    print(df_raw.describe().to_string())

    print("\n[OK] df_raw preserved untouched. Snapshot written to df_raw_snapshot.csv")
    print("[OK] Phase 1 complete — ready for Phase 2 (Data Quality & EDA).")
```
